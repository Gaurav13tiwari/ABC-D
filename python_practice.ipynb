```python
#  11 shorted list without using built in short function use < for ascending and >use for descending
lst =[34,12,89,56,29,11,98]

for i in range(len(lst)):
  for j in range(i+1,len(lst)):
    if lst[i]>lst[j]:
      lst[i],lst[j] = lst[j],lst[i]

print("sorted list:",lst)

```

```python
# factorial using function
def factorial(n) :
  fact = 1
  for i in range(1,n+1):
    fact = fact*i

  return fact

print(f"factorial of 5:,{factorial(5)}")
```

```python
# 6 count a vowels in string
def vowels_count(word) :
  vowels = "aeiouAEIOU"
  count = 0
  for i in word :
    if i in vowels:
      count = count +1
  return count

print(vowels_count("Python"))
```

```python
# find largest value in list
def largest(lst):
  max_value = lst[0]
  for i in lst:
    if i > max_value:
      max_value = i
  return max_value

print("max value :", largest([22,44,33,55,]))
```

```python
#  functoin to find 2nd largest value in list
def second_largest(A):
  unique = list(set(A))
  unique.sort()
  return unique[-2]

print("2n largest" ,second_largest([22,55,33]))
```

```python
# 11 count freq of item
def frequence(lst):
  freq = {}
  for i in lst:
    freq[i] = freq.get(i,0)+1
  return freq

print("freq of item:", frequence(["Apple","banana","Apple"]))
```

```python
# 12 check anagram
def is_anagram(n,m):
  return sorted(n) == sorted(m)

print("is anagram", is_anagram("listen", "silent"))
```

```python
# hourglass
n = 3
for i in range(n,0,-1):
  print(" "*(n-i)+"*"*(2*i-1))

for i in range(2,n+1):
  print(" "*(n-i) + "*"*(2*i-1))
```

```python
#  binary pattern
n = 5
for i in range(1,n+1):
  num = i%2
  for j in range(i):
    print(num,end=" ")
    num = 1-num
  print()
```
