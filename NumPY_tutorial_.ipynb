```python
angles = np.array([0,30,45,60,90,180])
print("given angles :", angles)

radian = np.deg2rad(angles)
print(radian)
```

```python
sin = np.sin(radian)
print(f"sin of angles{sin}")

# invers
sin_inverse = np.arcsin(radian)
print(f"inverse of sin:{sin_inverse}")

# hyperbolic
sin_h =np.sinh(radian)
print(f"hyperbolic sin: {sin_inverse}")

# invers sin_h
sinh_inverse = np.arcsinh(radian)
print(f"inverse hyperbolic sin : {sinh_inverse}")


```

stats func #

```python
array = np.array([22,33,4,4,55,11,])
print(f"given array:{array}")

print(f"numpy min :{np.min(array)} and numpy max:{np.max(array)}")
print(f"range of numpy: {np.ptp(array)}")  # diff of max min

print(f"percrntile:{np.percentile(array,30)}")

print(f"mean of array : {np.mean(array)}")

print(f"median of array : {np.median(array)}\n")
print(f"standard deviation of array :{np.std(array)}\n")

print(f"variance: {np.var(array)}")
print(f"avrege: {np.average(array)}")
```
