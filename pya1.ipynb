```python
# Validate Password
password = input("Enter password: ")

digit = False
special = False

for ch in password:
    if '0' <= ch <= '9':
        digit = True
    elif ch in "!@#$%^&*":
        special = True

if len(password) >= 8 and digit and special:
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
 # Check whether a number is prime
 n = int(input("Enter a number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count += 1

if count == 2:
    print("Prime")
else:
    print("Not Prime")
```

```python
# Check whether a number is a palindrome
n = int(input("Enter a number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Electricity bill calculation using slabs
units = int(input("Enter units: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = 100 * 5 + (units - 100) * 7
else:
    bill = 100 * 5 + 100 * 7 + (units - 200) * 10

print("Electricity Bill:", bill)
```

```python
# Income tax calculation based on slabs
income = int(input("Enter income: "))

if income <= 250000:
    tax = 0
elif income <= 500000:
    tax = (income - 250000) * 5 / 100
elif income <= 1000000:
    tax = 250000 * 5 / 100 + (income - 500000) * 20 / 100
else:
    tax = 250000 * 5 / 100 + 500000 * 20 / 100 + (income - 1000000) * 30 / 100

print("Tax:", tax)
```

```python
#  Check whether a number is an Armstrong number
n = int(input("Enter number: "))

original = n
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit * digit * digit
    n = n // 10

if sum == original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
# Fibonacci series
n = int(input("Enter N: "))

a = 0
b = 1

for i in range(n):
    print(a)
    c = a + b
    a = b
    b = c
```

```python
#  Find LCM and GCD of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

x = a
y = b

while y != 0:
    r = x % y
    x = y
    y = r

gcd = x
lcm = (a * b) // gcd

print("GCD:", gcd)
print("LCM:", lcm)
```

```python
# Function to return sum of digits
def sum_digits(n):
    sum = 0

    while n > 0:
        digit = n % 10
        sum = sum + digit
        n = n // 10

    return sum

n = int(input("Enter number: "))
print("Sum:", sum_digits(n))
```

```python
# Function to calculate power xⁿ without pow()
def power(x, n):
    result = 1

    for i in range(n):
        result = result * x

    return result

x = int(input("Enter x: "))
n = int(input("Enter n: "))

print("Power:", power(x, n))
```

# Interview-Level Function Questions

```python
# Function to find unique elements from a list
def unique(numbers):
    result = []

    for i in numbers:
        if i not in result:
            result.append(i)

    return result

numbers = [1, 2, 2, 3, 4, 4, 5]
print(unique(numbers))
```

```python
# Function to find duplicate elements in a list
def duplicates(numbers):
    result = []

    for i in numbers:
        if numbers.count(i) > 1 and i not in result:
            result.append(i)

    return result

numbers = [1, 2, 2, 3, 4, 4, 5]
print(duplicates(numbers))
```

```python
# Function to return frequency dictionary of list elements
def frequency(numbers):
    result = {}

    for i in numbers:
        if i in result:
            result[i] += 1
        else:
            result[i] = 1

    return result

numbers = [1, 2, 2, 3, 3, 3,4,2,4,5,5,3,5,4,3,2,]
print(frequency(numbers))
```

```python
# Function to check whether two strings are anagrams
def anagram(a, b):
    if sorted(a) == sorted(b):
        return True
    else:
        return False

a = input("Enter first string: ")
b = input("Enter second string: ")
print(anagram(a, b))
```

```python
# Function to flatten a nested list — Basic
def flatten(numbers):
    result = []

    for i in numbers:
        for j in i:
            result.append(j)

    return result

numbers = [[1, 2], [3, 4], [5, 6]]
print(flatten(numbers))
```

# Hiring-Round Combined Questions

```python
# Count even, odd, positive and negative numbers
def check(numbers):
    even = 0
    odd = 0
    positive = 0
    negative = 0

    for i in numbers:
        if i % 2 == 0:
            even += 1
        else:
            odd += 1

        if i > 0:
            positive += 1
        elif i < 0:
            negative += 1

    return even, odd, positive, negative

numbers = [2, -3, 4, -5, 7, 8, 9, -1, -6, 12, 77]

e, o, p, n = check(numbers)

print("Even:", e)
print("Odd:", o)
print("Positive:", p)
print("Negative:", n)
```

```python
# Function to return average and print grade
def marks_result(marks):
    total = 0

    for i in marks:
        total = total + i

    average = total / len(marks)

    print("Average:", average)

    if average >= 90:
        print("Grade: A")
    elif average >= 75:
        print("Grade: B")
    elif average >= 60:
        print("Grade: C")
    else:
        print("Grade: Fail")

marks = [85, 90, 78, 92, 88]
marks_result(marks)
```

```python
# Take input continuously, stop at 0 and print sum
sum = 0

while True:
    n = int(input("Enter number: "))

    if n == 0:
        break

    sum = sum + n

print("Sum:", sum)
```

```python
# Function to return longest word
def longest_word(sentence):
    words = sentence.split()
    longest = words[0]

    for w in words:
        if len(w) > len(longest):
            longest = w

    return longest

sentence = input("Enter sentence: ")
print("Longest word:", longest_word(sentence))
```

```python
# Menu-driven program using functions
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b


while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 5:
        print("Exit")
        break

    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    if choice == 1:
        print("Result:", add(a, b))
    elif choice == 2:
        print("Result:", subtract(a, b))
    elif choice == 3:
        print("Result:", multiply(a, b))
    elif choice == 4:
        print("Result:", divide(a, b))
    else:
        print("Invalid choice")
```
